```python
import yfinance as yf
import pandas as pd

# A sample of Nifty50 stocks across sectors + the Nifty50 index itself
tickers = ["RELIANCE.NS", "TCS.NS", "HDFCBANK.NS", "INFY.NS", "ITC.NS", "^NSEI"]

data = yf.download(tickers, start="2021-01-01", end="2026-09-01")["Close"]
data.head()
```

```python
returns = data.pct_change().dropna()
returns.head()
```

```python
import numpy as np

annual_return = returns.mean() * 252
annual_volatility = returns.std() * np.sqrt(252)

summary = pd.DataFrame({
    "Annual Return": annual_return,
    "Annual Volatility": annual_volatility
})
summary["Sharpe Ratio"] = summary["Annual Return"] / summary["Annual Volatility"]
summary
```

```python
import statsmodels.api as sm

market = returns["^NSEI"]
capm_results = {}

for stock in tickers[:-1]:  # exclude ^NSEI itself
    y = returns[stock]
    X = sm.add_constant(market)
    model = sm.OLS(y, X).fit()
    alpha, beta = model.params
    capm_results[stock] = {"Alpha": alpha, "Beta": beta, "R-squared": model.rsquared}

capm_df = pd.DataFrame(capm_results).T
capm_df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(summary["Annual Volatility"][:-1], summary["Annual Return"][:-1])

for stock in summary.index[:-1]:
    plt.annotate(stock, (summary["Annual Volatility"][stock], summary["Annual Return"][stock]))

plt.scatter(summary["Annual Volatility"]["^NSEI"], summary["Annual Return"]["^NSEI"], color="red", label="Nifty50 Index")
plt.annotate("^NSEI", (summary["Annual Volatility"]["^NSEI"], summary["Annual Return"]["^NSEI"]))

plt.xlabel("Annual Volatility (Risk)")
plt.ylabel("Annual Return")
plt.title("Risk vs Return: Nifty50 Stocks vs Index")
plt.legend()
plt.grid(True)
plt.show()
```
